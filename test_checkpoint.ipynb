# Checkpoint Test Notebook
Tests kernel state persistence, editor state, widgets, and matplotlib.

```python
import numpy as np
x = np.linspace(0, 10, 100)
y = np.sin(x)
state_var = 'checkpoint_test_value'
print(f'State variable set: {state_var}')
```

```python
%matplotlib inline
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, label='sin(x)')
ax.plot(x, np.cos(x), label='cos(x)')
ax.set_title('Matplotlib test')
ax.legend()
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display

slider = widgets.IntSlider(value=42, min=0, max=100, description='Test slider:')
display(slider)
```

```python
text = widgets.Text(value='hello checkpoint', description='Text:')
dropdown = widgets.Dropdown(options=['alpha', 'beta', 'gamma'], value='beta', description='Pick:')
output = widgets.Output()

def on_change(change):
    with output:
        output.clear_output()
        print(f'slider={slider.value}, text={text.value}, dropdown={dropdown.value}')

slider.observe(on_change, names='value')
text.observe(on_change, names='value')
dropdown.observe(on_change, names='value')

display(widgets.VBox([text, dropdown, output]))
```
